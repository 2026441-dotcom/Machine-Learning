```python
X = df.drop(columns=["G1", "G2", "G3"])
y = df["G3"]

print(X.shape)
print(y.shape)
```

```python
X = pd.get_dummies(X, drop_first=True, dtype=int)

print(X.shape)
X.head()
```

### Correlation Analysis

```python
correlation = df.select_dtypes(include="number").corr()

print(correlation["G3"].sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt

correlation["G3"].drop("G3").sort_values().plot(
    kind="barh",
    figsize=(9, 6)
)

plt.title("Correlation of Features with Final Grade (G3)")
plt.xlabel("Correlation")
plt.ylabel("Features")
plt.axvline(0, color="black")
plt.tight_layout()
plt.show()
```

Correlation Analysis – Interpretation

From the results, I can see that G2 and G1 have the strongest positive relationship with the final grade (G3). G2 has a correlation of 0.92 and G1 has 0.83. Study time also has a positive relationship, but it is much weaker. Failures have a negative relationship with G3 (-0.39). I think previous grades are very useful for predicting students' final performance. This also supports my Random Forest feature importance results.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

Why I Chose These Models

I chose Linear Regression and Random Forest because I want to predict students' final grades (G3).

I chose Linear Regression because it is simple and easy to understand. Also, G3 is a numerical value, so I think this model is suitable for my project.

I chose Random Forest because my dataset has different information about students, such as study time, absences and previous grades. I wanted to see if this model could find more patterns in the data.

I decided to use these two models because I wanted to compare their results and see which one can predict students' final grades better.

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

y_pred = lr_model.predict(X_test)

print("Model trained successfully")
```

```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mse)
print("R2 Score:", r2)
```

The Linear Regression model has an MSE of 8.24 and an R² score of 0.18. This means the model explains about 18% of the differences in students' final grades. I think the result is quite low, so the model is not very accurate at predicting final grades using these features.

```python
for size in [0.05, 0.15, 0.25]:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=size, random_state=42
    )

    lr_model = LinearRegression()
    lr_model.fit(X_train, y_train)

    y_pred = lr_model.predict(X_test)

    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    print("Test Size:", size)
    print("MSE:", round(mse, 2))
    print("R2 Score:", round(r2, 2))
    print("--------------------")
```

I tested the Linear Regression model using 5%, 15%, and 25% test sizes. The 5% test size had the lowest MSE of 7.42, while the 15% test size had the highest R² score of 0.22. The 25% test size gave slightly weaker results.

I think the model's performance is not very good, and changing the test size did not improve it much. Because of this, I decided to include the students' previous grades (G1 and G2) and compare the results to see if the model could perform better.

```python
X_with_grades = df.drop(columns=["G3"])
X_with_grades = pd.get_dummies(X_with_grades, drop_first=True, dtype=int)

print(X_with_grades.shape)
```

```python
for size in [0.05, 0.15, 0.25]:
    X_train, X_test, y_train, y_test = train_test_split(
        X_with_grades, y, test_size=size, random_state=42
    )

    lr_model = LinearRegression()
    lr_model.fit(X_train, y_train)

    y_pred = lr_model.predict(X_test)

    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    print("Test Size:", size)
    print("MSE:", round(mse, 2))
    print("R2 Score:", round(r2, 2))
    print("--------------------")
```

After adding G1 and G2, the Linear Regression model performed much better. The R² scores increased to 0.94, 0.84, and 0.87 for the three test sizes. I think previous grades are very important for predicting final grades. However, the 5% test set is quite small, so its high score may not be as reliable as the results from larger test sets.

```python
from sklearn.ensemble import RandomForestRegressor

X_train, X_test, y_train, y_test = train_test_split(
    X_with_grades, y, test_size=0.25, random_state=42
)

rf_model = RandomForestRegressor(random_state=42)
rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)

print("Random Forest model trained successfully")
```

```python
mse_rf = mean_squared_error(y_test, y_pred_rf)
r2_rf = r2_score(y_test, y_pred_rf)

print("Random Forest MSE:", round(mse_rf, 2))
print("Random Forest R2 Score:", round(r2_rf, 2))
```

The Random Forest model gave an MSE of 1.47 and an R² score of 0.85. This is a good result, but Linear Regression performed slightly better with an MSE of 1.36 and an R² score of 0.87. I think both models can predict students' final grades quite well when G1 and G2 (previous period grades) are included.

```python
for size in [0.05, 0.15, 0.25]:
    X_train, X_test, y_train, y_test = train_test_split(
        X_with_grades, y, test_size=size, random_state=42
    )

    rf_model = RandomForestRegressor(random_state=42)
    rf_model.fit(X_train, y_train)

    y_pred_rf = rf_model.predict(X_test)

    mse_rf = mean_squared_error(y_test, y_pred_rf)
    r2_rf = r2_score(y_test, y_pred_rf)

    print("Test Size:", size)
    print("MSE:", round(mse_rf, 2))
    print("R2 Score:", round(r2_rf, 2))
    print("--------------------")
```

I tested the Random Forest model using 5%, 15%, and 25% test sizes. The R² scores were 0.92, 0.81, and 0.85. After comparing the results with Linear Regression, I noticed that Linear Regression performed slightly better in all three test sizes. However, both models gave good results when G1 and G2 (previous period grades) were included. I think Linear Regression is the better model for this dataset based on these results.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_with_grades, y, test_size=0.25, random_state=42
)

lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
y_pred_lr = lr_model.predict(X_test)

r2_lr = r2_score(y_test, y_pred_lr)

plt.scatter(y_test, y_pred_lr, s=10)
plt.plot([0, 20], [0, 20], color="red")

plt.title(f"Linear Regression (R² = {r2_lr:.2f})")
plt.xlabel("Actual Final Grade (G3)")
plt.ylabel("Predicted Final Grade (G3)")
plt.show()
```

The graph compares the actual final grades with the grades predicted by Linear Regression. Most of the blue points are close to the red line, which means the model predicts the grades quite well. The R² score is 0.87, which is a good result. However, some points are far from the line, so the model does not predict every student's grade correctly. I think adding G1 and G2 helped improve the predictions.

```python
rf_model = RandomForestRegressor(random_state=42)
rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)
r2_rf = r2_score(y_test, y_pred_rf)

plt.scatter(y_test, y_pred_rf, s=10)
plt.plot([0, 20], [0, 20], color="red")

plt.title(f"Random Forest (R² = {r2_rf:.2f})")
plt.xlabel("Actual Final Grade (G3)")
plt.ylabel("Predicted Final Grade (G3)")
plt.show()
```

```python
from sklearn.model_selection import cross_val_score
import numpy as np

lr_model = LinearRegression()

cv_scores = cross_val_score(
    lr_model, X_with_grades, y, cv=10, scoring='r2'
)

mean_cv_score = np.mean(cv_scores)

print(f"Mean Cross-Validation Score (R²): {mean_cv_score:.4f}")
```

I used 10-fold cross-validation to check the Linear Regression model. The average R² score was 0.80, which is a good result. It is slightly lower than the previous test score of 0.87, but I think the model still performs well across different parts of the dataset.

```python
rf_model = RandomForestRegressor(random_state=42)

cv_scores_rf = cross_val_score(
    rf_model, X_with_grades, y, cv=10, scoring='r2'
)

mean_cv_rf = np.mean(cv_scores_rf)

print(f"Mean Cross-Validation Score (R²): {mean_cv_rf:.4f}")
```

```python
models = ["Linear Regression", "Random Forest"]
scores = [0.80, 0.78]

plt.bar(models, scores)
plt.title("10-Fold Cross-Validation Comparison")
plt.ylabel("Average R² Score")
plt.ylim(0, 1)
plt.show()
```

I used 10-fold cross-validation to test the Random Forest model and got an average R² score of 0.78. Linear Regression scored 0.80, which is slightly higher. I think both models performed well, but Linear Regression gave better results in this comparison.

```python
from sklearn.model_selection import cross_val_score

for size in [0.05, 0.15, 0.25]:

    X_train, X_test, y_train, y_test = train_test_split(
        X_with_grades, y, test_size=size, random_state=42
    )

    lr_cv = cross_val_score(
        LinearRegression(),
        X_train, y_train,
        cv=10, scoring="r2"
    )

    rf_cv = cross_val_score(
        RandomForestRegressor(random_state=42),
        X_train, y_train,
        cv=10, scoring="r2"
    )

    print("Test Size:", size)
    print("Linear Regression CV R2:", round(lr_cv.mean(), 2))
    print("Random Forest CV R2:", round(rf_cv.mean(), 2))
    print()
```

```python
import matplotlib.pyplot as plt

test_sizes = ["5%", "15%", "25%"]
lr_scores = [0.84, 0.84, 0.82]
rf_scores = [0.84, 0.85, 0.83]

plt.figure(figsize=(8, 5))

plt.plot(test_sizes, lr_scores, marker="o", label="Linear Regression")
plt.plot(test_sizes, rf_scores, marker="o", label="Random Forest")

plt.title("Cross-Validation Results for Different Test Sizes")
plt.xlabel("Test Size")
plt.ylabel("Average CV R² Score")
plt.ylim(0.75, 0.90)
plt.legend()
plt.grid(True)
plt.show()
```

Cross-Validation for Different Test Sizes

I used 10-fold cross-validation with three different test sizes: 5%, 15%, and 25%. The R² scores were between 0.82 and 0.85 for both models. I can see that the results are quite similar across the different test sizes. I think both models performed consistently, and there was no big difference between them.

```python
from sklearn.model_selection import GridSearchCV

rf_model = RandomForestRegressor(random_state=42)

param_grid = {
    "n_estimators": [50, 100],
    "max_depth": [5, 10, None]
}

grid_search = GridSearchCV(
    rf_model,
    param_grid,
    cv=5,
    scoring="r2"
)

grid_search.fit(X_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best CV R2:", round(grid_search.best_score_, 2))
```

GridSearchCV Results

I used GridSearchCV to find better settings for my Random Forest model. I tested different numbers of trees (50 and 100) and different maximum depths (5, 10 and None). I used 5-fold cross-validation to compare these settings.

From the results, I found that 50 trees and a maximum depth of 5 gave the best result, with a cross-validation R² score of 0.84. I chose these settings because they performed the best among the options I tested.

```python
best_rf = grid_search.best_estimator_

y_pred_best_rf = best_rf.predict(X_test)

mse_best_rf = mean_squared_error(y_test, y_pred_best_rf)
r2_best_rf = r2_score(y_test, y_pred_best_rf)

print("Tuned Random Forest MSE:", round(mse_best_rf, 2))
print("Tuned Random Forest R2:", round(r2_best_rf, 2))
```

```python
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

train_pred = lr_model.predict(X_train)
test_pred = lr_model.predict(X_test)

print("Training R2:", round(r2_score(y_train, train_pred), 2))
print("Testing R2:", round(r2_score(y_test, test_pred), 2))
```

I compared the training and testing R² scores to check for overfitting. The training score was 0.85 and the testing score was 0.87. The difference is very small, so I think the Linear Regression model is performing well on both datasets, with no clear signs of overfitting or underfitting.

```python
rf_train_pred = best_rf.predict(X_train)
rf_test_pred = best_rf.predict(X_test)

print("Random Forest Training R2:", round(r2_score(y_train, rf_train_pred), 2))
print("Random Forest Testing R2:", round(r2_score(y_test, rf_test_pred), 2))
```

```python
import matplotlib.pyplot as plt
import numpy as np

models = ["Linear Regression", "Random Forest"]

train_scores = [0.85, 0.92]
test_scores = [0.87, 0.86]

x = np.arange(len(models))
width = 0.35

plt.figure(figsize=(8, 5))

plt.bar(x - width/2, train_scores, width, label="Training R2")
plt.bar(x + width/2, test_scores, width, label="Testing R2")

plt.xticks(x, models)
plt.ylabel("R2 Score")
plt.title("Training vs Testing R2 Scores")
plt.ylim(0, 1)
plt.legend()

plt.show()
```

I compared the training and testing R² scores for both models. Linear Regression achieved 0.85 on training data and 0.87 on testing data, showing no clear signs of overfitting. Random Forest achieved 0.92 on training data and 0.86 on testing data. This suggests that Random Forest may have slight overfitting, but it still performs well on the testing data. Overall, both models gave good results.
